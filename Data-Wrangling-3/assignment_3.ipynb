```python
df = pd.read_csv("nba.csv")
df
```

```python
df.isnull().sum()
```

```python
df["College"] = df["College"].fillna(df["College"].mode()[0])
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace = True)
```

```python
df.isnull().sum()
```

```python
age_groups = df.groupby(df["Age"])
age_groups.head()
```

```python
age_groups["Salary"].describe()
```
